## 3. Variáveis: dando nome aos valores

Uma variável é uma **etiqueta** colada em um valor. O sinal `=` significa "guarde": o valor da direita vai para o nome da esquerda.

````python
preco = 39.90
meses = 12
total = preco * meses
print(total)
````

**Por que `478.79999999999995` e não `478.8`?** O computador guarda decimais em binário, e alguns números não cabem exatamente, assim como 1/3 não cabe em decimal. A diferença é minúscula e some quando formatamos o número para exibição, como veremos na seção 5.

````python
# Uma variável pode receber um novo valor; o antigo é esquecido
meses = 6
total = preco * meses
print(total)
````

**Regras para nomes**

- letras minúsculas, números e `_`; sem espaços e sem acentos: `preco_unitario`
- não pode começar com número: `2_semestre` é inválido
- maiúsculas e minúsculas são diferentes: `Total` e `total` são variáveis distintas
- escolha nomes que expliquem o conteúdo: `dias_de_viagem` é melhor do que `d`

## 8. Prática em subgrupos: calculadora de orçamento de viagem

**Formato:** salas de até 5 pessoas · 30 minutos · uma pessoa compartilha a tela e digita, as demais orientam.

**Missão:** o programa pergunta os dados de uma viagem em grupo e mostra o orçamento.

Entradas (use `input` e converta os tipos):

1. número de dias
2. número de pessoas
3. valor da diária do hotel (o quarto é dividido por todos)
4. gasto diário com alimentação **por pessoa**
5. valor da passagem **por pessoa**

Saídas:

- custo de hospedagem, alimentação e transporte do grupo
- reserva de emergência: 10% da soma
- total geral e valor por pessoa, com duas casas decimais

Exemplo de relatório:

```
===== ORÇAMENTO DA VIAGEM =====
Hospedagem:   R$ 1100.00
Alimentação:  R$ 1800.00
Transporte:   R$ 3400.00
Reserva (10%): R$ 630.00
-------------------------------
TOTAL:        R$ 6930.00
Por pessoa:   R$ 1732.50
```

Ao voltar para a sala principal, a pessoa porta-voz conta: qual erro apareceu e como o grupo resolveu?

````python
# Prática: calculadora de orçamento de viagem
# escreva seu código aqui
dias = int(input('N de dias: '))
pessoas = int(input('N de pessoas: '))
diaria = float(input('R$ da diaria: '))
alimentacao = float(input('R$ alimentação diario por pessoa: '))
passagem = float(input('R$ da passagem por pessoa: '))

hospedagem = diaria * dias
alimentacao = alimentacao * dias * pessoas
transporte = passagem * pessoas
total = hospedagem + alimentacao + transporte
reserva = (10/100) * total

total_geral = total + reserva
por_pessoa = total_geral/pessoas
print('==== ORÇAMENTO DA VIAGEM====')
print(f'Hospedagem:    R$ {hospedagem:.2f}')
print(f'Alimentação:   R$ {alimentacao:.2f}')
print(f'Transporte:    R$ {transporte:.2f}')
print(f'Reserva (10%): R$ {reserva:.2f}')
print('----------------------------')
print(f'TOTAL:         R$ {total_geral:.2f}')
print(f'Por pessoa:    R$ {por_pessoa:.2f}')


````

### Reforço 9.3 [Reforço]
Um caixa eletrônico precisa entregar R$ 390 usando o menor número de notas de 100, 50 e 20. Use `//` e `%` para descobrir quantas notas de cada valor. Confira: o resultado deve somar 390.

### Desafio 9.4 [Desafio]
Juros compostos: quanto rendem R$ 5.000 aplicados por 3 anos a 0,8% ao mês? Fórmula: `montante = capital * (1 + taxa) ** meses`. Mostre o montante e o total de juros.

````python
# Desafio 9.4
# escreva seu código aqui
capital = 5000
meses = 3 * 12
taxa = 0.8/100
montante = capital * (1+taxa) ** meses
print(f"Rendem R$ {montante:.2f}")
print(f"Total de juros R$ {montante - capital:.2f}")
````
