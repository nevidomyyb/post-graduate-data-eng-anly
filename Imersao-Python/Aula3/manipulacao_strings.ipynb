## 1. Texto é uma sequência de caracteres

Cada caractere tem uma posição (índice), contada a partir de **zero**. Índices negativos contam do fim.

```
 texto:      P   y   t   h   o   n
 índices:    0   1   2   3   4   5
 negativos: -6  -5  -4  -3  -2  -1
```

````python
palavra = "Python"
print(palavra[0])     # primeiro caractere
print(palavra[-1])    # último
print(len(palavra))   # quantidade de caracteres
````

### Fatias: `texto[inicio:fim:passo]`

O `fim` **não entra**. Se omitir o início, começa do zero; se omitir o fim, vai até o final.

````python
palavra = "Python"
print(palavra[0:2])    # Py
print(palavra[2:])     # thon
print(palavra[:-1])    # tudo menos o último
print(palavra[::2])    # de 2 em 2: Pto
print(palavra[::-1])   # passo negativo: texto invertido
````

Strings são **imutáveis**: não dá para trocar uma letra no lugar. Execute e leia o erro.

````python
palavra = "Python"
palavra[0] = "J"
````

````python
# Em vez de alterar, criamos um texto novo
palavra = "Python"
nova = "J" + palavra[1:]
print(nova)
````

## 2. Limpando e padronizando

| Método | O que faz | Exemplo | Resultado |
| --- | --- | --- | --- |
| `upper()` | tudo maiúsculo | `"ana".upper()` | `"ANA"` |
| `lower()` | tudo minúsculo | `"ANA".lower()` | `"ana"` |
| `title()` | inicial maiúscula em cada palavra | `"ana lima".title()` | `"Ana Lima"` |
| `capitalize()` | só a primeira letra do texto | `"ana lima".capitalize()` | `"Ana lima"` |
| `strip()` | remove espaços das pontas | `"  ana ".strip()` | `"ana"` |
| `lstrip()` / `rstrip()` | só da esquerda / só da direita | `"  ana ".rstrip()` | `"  ana"` |
| `replace(a, b)` | troca `a` por `b` | `"a-b".replace("-", "/")` | `"a/b"` |

**Importante:** métodos de texto **devolvem um texto novo**. A variável original só muda se você guardar o resultado nela.

````python
nome = "  mARIA lima  "
nome.strip()          # devolve um texto novo...
print(f"[{nome}]")    # ...mas nome continua igual

nome = nome.strip()   # agora sim, guardamos o resultado
print(f"[{nome}]")
````

````python
# Métodos podem ser encadeados: executam da esquerda para a direita
bruto = "  mARIA lima  "
limpo = bruto.strip().lower().title()
print(limpo)
````

````python
# replace pode receber um terceiro argumento: quantas trocas fazer
texto = "banana"
print(texto.replace("a", "o"))
print(texto.replace("a", "o", 1))
````

## 3. Procurando dentro do texto

| Ferramenta | O que responde | Exemplo |
| --- | --- | --- |
| `a in texto` | `a` aparece no texto? (`True`/`False`) | `"@" in email` |
| `texto.count(a)` | quantas vezes `a` aparece | `"banana".count("a")` → `3` |
| `texto.find(a)` | posição da primeira ocorrência (ou `-1`) | `"banana".find("n")` → `2` |
| `startswith(a)` / `endswith(a)` | começa / termina com `a`? | `arquivo.endswith(".csv")` |
| `isdigit()` / `isalpha()` | só dígitos? só letras? | `"2026".isdigit()` → `True` |
| `isupper()` / `islower()` | tudo maiúsculo? tudo minúsculo? | `"ABC".isupper()` → `True` |

````python
email = "ana.souza@empresa.com.br"
print("@" in email)
print(email.count("."))
print(email.find("@"))
print(email.find("#"))            # não existe: -1
print(email.endswith(".br"))
print(email[:email.find("@")])    # tudo antes do @
````

Uma lista se acessa por índice, igualzinho a um texto: `partes[0]`, `partes[-1]`. Vamos estudar listas com calma na Aula 2; por enquanto, é tudo o que você precisa saber.

````python
linha = "Ana;32;Recife;Saúde"
partes = linha.split(";")
print(partes)
print(partes[0])
idade = int(partes[1])    # cada pedaço é texto: converta quando precisar
print(idade + 1)
````

## 5. Formatando a saída com f-strings

Depois dos dois-pontos, dentro das chaves, vem a **especificação de formato**:

| Formato | Significado | Exemplo | Resultado |
| --- | --- | --- | --- |
| `:.2f` | 2 casas decimais | `f"{3.14159:.2f}"` | `3.14` |
| `:,.2f` | separador de milhar | `f"{1234567.8:,.2f}"` | `1,234,567.80` |
| `:<10` | alinha à esquerda em 10 posições | `f"[{'ab':<10}]"` | `[ab        ]` |
| `:>10` | alinha à direita | `f"[{'ab':>10}]"` | `[        ab]` |
| `:^10` | centraliza | `f"[{'ab':^10}]"` | `[    ab    ]` |
| `:05d` | inteiro com zeros à esquerda | `f"{42:05d}"` | `00042` |
| `:.1%` | percentual | `f"{0.256:.1%}"` | `25.6%` |

Outros recursos úteis: `"=" * 30` repete um caractere; `\n` pula linha; `\t` insere tabulação.

````python
produto = "Caderno"
preco = 18.9
print(f"[{produto:<12}]")
print(f"[{produto:>12}]")
print(f"[{produto:^12}]")
print(f"{produto:<12}R$ {preco:>8.2f}")
print("=" * 23)
print(f"Pedido nº {42:05d}\nTaxa: {0.0875:.2%}")
````

````python
# Formato brasileiro de moeda: o Python usa o padrão americano; um truque com replace resolve
valor = 1234567.891
americano = f"{valor:,.2f}"
brasileiro = americano.replace(",", "X").replace(".", ",").replace("X", ".")
print(americano)
print(f"R$ {brasileiro}")
````

## 7. Prática em subgrupos: padronizando um cadastro

**Formato:** salas de até 5 pessoas · 25 minutos · uma pessoa compartilha a tela e digita, as demais orientam.

Um formulário on-line gravou os dados sem nenhum cuidado. Cada registro é uma linha com os campos `nome;cpf;email;telefone;cidade-uf`.

**Missão:** transformar o registro bruto em uma ficha padronizada:

- **Nome:** espaços normalizados, iniciais maiúsculas, preposições (`da`, `de`, `do`, `dos`, `das`) em minúsculas;
- **CPF** e **Telefone:** só dígitos;
- **E-mail:** minúsculas, sem espaços;
- **Cidade/UF:** cidade com iniciais maiúsculas, UF em maiúsculas, separadas por `/`.

Por fim, monte a **linha padronizada** juntando os campos com `;` (use `join`).

Quando terminar, troque o valor de `registro` pelos outros dois exemplos (comentados) e confira se o seu código continua funcionando.

### Prática 7.1 [Essencial]
Monte a ficha padronizada.

Saída esperada:

```
======= FICHA PADRONIZADA =======
Nome:      José Carlos da Silva
CPF:       09198765432
E-mail:    jose.carlos@email.com
Telefone:  91988887777
Cidade/UF: Belém/PA
=================================
José Carlos da Silva;09198765432;jose.carlos@email.com;91988887777;Belém/PA
```

````python
# Prática 7.1
# registro = "  jOSÉ carlos   DA   silva ; 091.987.654-32 ;  JOSE.CARLOS@EMAIL.COM ;(91)98888-7777; belém-pa "
registro = "ana PAULA dos   reis;123.456.789-00;Ana.Reis@Empresa.com.br ; (11) 3333-4444 ;são paulo-SP"
# registro = "   MARCOS de oliveira;  987.654.321-00;marcos@uol.com.br;(81) 99999-0000  ;RECIFE-pe"
# escreva seu código aqui
nome, cpf, email, telefone, cidade = registro.split(";")
nome = nome.strip().title()
preposicoes = ["Da", "De", "Do", "Dos", "Das"]
for prep in preposicoes:
    nome = nome.replace(prep, prep.lower())
nome = nome.split()
nome = " ".join(nome)
cpf = cpf.strip().replace(".", "").replace("-", "")
email = email.lower().strip()
telefone = telefone.strip().replace("(", "").replace(")", "").replace("-", "")
cidade, uf = cidade.strip().split("-")
cidade = cidade.title()
uf = uf.upper()
print("======= FICHA PADRONIZADA =======") #35
print(f"{'Nome:':<11}{nome}")
print(f"{'CPF:':<11}{cpf}")
print(f"{'E-mail:':<11}{email}")
print(f"{'Telefone:':<11}{telefone}")
print(f"{'Cidade/UF:':<11}{cidade}/{uf}")
print("=================================")
itens = [nome, cpf, email, telefone, f"{cidade}/{uf}"]
registro = ";".join(itens)
print(registro)
````

### Desafio 8.5 [Desafio]
Uma URL tem várias partes. Extraia, só com métodos de texto e fatias:

1. o protocolo (antes de `://`);
2. o domínio (entre `://` e a primeira `/` depois dele);
3. o caminho (da primeira `/` depois do domínio até o `?`);
4. o valor do parâmetro `pagina`.

Dica: `find` aceita um segundo argumento, a posição a partir da qual procurar: `texto.find("/", 8)`.

Saída esperada:

```
Protocolo: https
Domínio:   www.exemplo.com.br
Caminho:   /cursos/python
Página:    3
```

````python
# Desafio 8.5
url = "https://www.exemplo.com.br/cursos/python?pagina=3"

# 1. Protocolo
pos_proto = url.find("://")
protocolo = url[:pos_proto]

inicio_dom = pos_proto + 3
fim_dom = url.find("/", inicio_dom)
dominio = url[inicio_dom:fim_dom]

fim_caminho = url.find("?", fim_dom)
caminho = url[fim_dom:fim_caminho]

texto_param = "pagina="
pos_param = url.find(texto_param)
pagina = url[pos_param + len(texto_param):]

print(f"Protocolo: {protocolo}")
print(f"Domínio:   {dominio}")
print(f"Caminho:   {caminho}")
print(f"Página:    {pagina}")
````

## Resumo

- Índices começam em 0; `texto[-1]` é o último caractere; `texto[inicio:fim]` não inclui o fim; `texto[::-1]` inverte.
- Strings são imutáveis: métodos **devolvem um texto novo**, que precisa ser guardado.
- Limpeza: `strip`, `lower`, `upper`, `title`, `replace`. A ordem das trocas importa.
- Busca: `in`, `count`, `find`, `startswith`, `endswith`, `isdigit`, `isalpha`.
- `split` quebra e `join` junta; `" ".join(texto.split())` normaliza espaços.
- f-strings formatam: `:.2f`, `:,.2f`, `:<10`, `:>10`, `:^10`, `:05d`, `:.1%`.
