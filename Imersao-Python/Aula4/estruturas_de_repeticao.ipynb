## 2. Ferramentas do `for`: `enumerate`, `zip`, `break` e `continue`

| Ferramenta | Para que serve |
| --- | --- |
| `enumerate(seq)` | entrega a **posição e o item** a cada volta |
| `zip(a, b)` | percorre duas sequências **em paralelo** |
| `break` | **interrompe** o laço imediatamente |
| `continue` | **pula** para a próxima volta |

````python
tarefas = ["limpar dados", "gerar gráfico", "enviar relatório"]
for i, tarefa in enumerate(tarefas, start=1):
    print(f"{i}. {tarefa}")
````

````python
produtos = ["Caneta", "Caderno", "Mochila"]
precos = [3.5, 18.9, 129.0]
for produto, preco in zip(produtos, precos):
    print(f"{produto:<10} R$ {preco:>7.2f}")
````

````python
# break: para quando encontrar; continue: pula o que não interessa
codigos = ["A10", "", "B22", "X99", "C31"]
for codigo in codigos:
    if codigo == "":
        continue          # ignora vazios
    if codigo.startswith("X"):
        print("Código proibido encontrado; parando.")
        break
    print(f"Processando {codigo}")
````

1. normalize cada um (sem espaços nas pontas, minúsculo);

3. separe em `validos` e `invalidos`, usando as regras da Prática extra 2 (sem espaço no meio, exatamente um `@`, não começa nem termina com `@`, domínio com ponto);
4. não deixe e-mails **repetidos** entrarem em `validos`.

Mostre as duas listas e as quantidades.

Saída esperada:

```
Válidos (3): ['ana@empresa.com', 'rui@uol.com.br', 'lia.m@dados.org']
Inválidos (4): ['joao@@mail.com', 'carla mendes@gmail.com', 'bia@empresa', '@site.com']
```

````python
# Sua vez 2.3
brutos = ["  Ana@Empresa.com ", "", "joao@@mail.com", "ANA@empresa.com", "carla mendes@gmail.com",
          "rui@uol.com.br", "   ", "bia@empresa", "@site.com", "Rui@UOL.com.br", "lia.m@dados.org"]
# escreva seu código aqui\
validos = []
invalidos = []

for email in brutos:
    e = email.strip().lower()

    if not e:
        continue

    tem_espaco = " " in e
    um_arroba = e.count("@") == 1
    ponta_arroba = e.startswith("@") or e.endswith("@")
    dominio_ponto = um_arroba and "." in e.split("@")[1]

    if not tem_espaco and um_arroba and not ponta_arroba and dominio_ponto:
        if e not in validos:
            validos.append(e)
    else:
        invalidos.append(e)

print(f"Válidos ({len(validos)}): {validos}")
print(f"Inválidos ({len(invalidos)}): {invalidos}")

    

    
````

## 3. Contando e agrupando com dicionários

Lembra do padrão de contagem da Aula 2? Ele é a base de muita análise de texto:

```python
contagem[chave] = contagem.get(chave, 0) + 1
```

````python
palavra = "abracadabra"
freq = {}
for letra in palavra:
    freq[letra] = freq.get(letra, 0) + 1
print(freq)

# Ordenando as chaves pelo valor, do maior para o menor
ordenadas = sorted(freq, key=freq.get, reverse=True)
print(ordenadas)
````

Na linha do `sorted`, o `key=freq.get` diz ao Python: "para ordenar cada chave, use o valor que ela tem no dicionário". É um atalho útil que vamos reutilizar abaixo.

### Sua vez 3.1 [Essencial] · Palavras mais frequentes
Descubra as **3 palavras mais frequentes** do texto:

1. converta para minúsculas;
2. remova a pontuação `.,;:!?` (dica: um laço sobre `".,;:!?"` com `replace`);
3. quebre em palavras e conte com um dicionário, **ignorando** as palavras da lista `stopwords`;
4. mostre as 3 primeiras no formato do exemplo.

Saída esperada:

```
1º dados: 3
2º geram: 2
3º análises: 2
```

````python
# Sua vez 3.1
texto = """Dados limpos geram boas análises. Dados sujos geram análises ruins!
Por isso, limpar dados é a etapa mais importante: sem limpeza, não há análise."""
stopwords = ["a", "o", "e", "é", "de", "da", "do", "por", "isso", "mais", "sem", "não", "há", "boas"]
# escreva seu código aqui
texto = texto.lower()
for i in ".,;:!?":
    texto = texto.replace(i, "")
palavras = texto.split()
_dict = {}
for palavra in palavras:
    if palavra not in stopwords:
        _dict[palavra] = _dict.get(palavra, 0) + 1


ordered_keys = sorted(_dict, key=_dict.get, reverse=True)
for i, key in enumerate(ordered_keys):
    if i <= 2:
        print(f"{i+1} {key}: {_dict[key]}")
````

### Sua vez 3.3 [Reforço] · Anagramas
Duas palavras são anagramas quando usam exatamente as mesmas letras, na mesma quantidade (ex.: `amor` e `roma`). Ignorando maiúsculas e espaços, verifique cada par da lista **contando as letras com dicionários** (não use `sorted`).

Saída esperada:

```
Amor / Roma: são anagramas
dados / sodad: são anagramas
python / typhoon: não são anagramas
Iracema / America: são anagramas
```

````python
# Sua vez 3.3
pares = [("Amor", "Roma"), ("dados", "sodad"), ("python", "typhoon"), ("Iracema", "America")]
# escreva seu código aqui
for p1, p2 in pares:
    q1 = {}
    q2 = {}
    
    for c in p1.lower().replace(" ", ""):
        q1[c] = q1.get(c, 0) + 1
        
    for c in p2.lower().replace(" ", ""):
        q2[c] = q2.get(c, 0) + 1
        
    if q1 == q2:
        print(f"{p1} / {p2}: são anagramas")
    else:
        print(f"{p1} / {p2}: não são anagramas")
        
````

## Resumo

- `for` percorre textos (caractere a caractere) e listas; com `split()`, percorre palavras.
- Acumuladores: `0` para somas e contagens, `""` para textos, `[]` para listas, `{}` para contagens por chave. **Sempre inicialize antes do laço.**
- `enumerate` dá posição e item; `zip` percorre em paralelo; `break` interrompe; `continue` pula.
- `contagem[chave] = contagem.get(chave, 0) + 1` conta qualquer coisa; `sorted(d, key=d.get, reverse=True)` ordena pela contagem.
- `while` repete enquanto a condição for verdadeira; garanta que algo mude a cada volta. `while True` + `break` é ótimo para validar entradas.
- Não altere uma lista enquanto a percorre: monte uma nova.
- Limpar → quebrar → contar é a base da análise de texto, e é o que o pandas fará por você a partir da Aula 4.
