# 6) Métodos matemáticos e estatísticos

```python
# Axis = 0 -> operação nas linhas
# Axis = 1 -> operação nas colunas

print(f"Soma das linhas: {m.sum(axis=0)}")
print(f"Soma das colunas: {m.sum(axis=1)}")
print(f"\nSoma cumulativa: {m.cumsum()}")
print(f"\nSoma cumulativa das colunas: {m.cumsum(axis=1)}")
```

# 11) Análise de dados com NumPy    

```python
import matplotlib.pyplot as plt

temperaturas = np.array([
    30.5, 32.1, 28.7, 29.8, 30.9, 31.6
])
dias = np.arange(1, len(temperaturas)+1)
print(temperaturas, dias)

max_temp = temperaturas.max()
min_temp = temperaturas.min()
mean_temp = temperaturas.mean().round(2)
days_above_mean = dias[temperaturas > mean_temp]
print(f'Máxima: {max_temp}')
print(f'Mínima: {min_temp}')
print(f'Média: {mean_temp}')
print(f'Dias acima da média: {days_above_mean}')

plt.plot(dias, temperaturas)
plt.scatter(dias[temperaturas.argmax()], max_temp, color='red', label='Máxima')
plt.scatter(dias[temperaturas.argmin()], min_temp, color='purple', label='Mínima')
plt.axhline(mean_temp, color='grey', linestyle='--', label="Média")
plt.xlabel("Dias")
plt.ylabel("Temperaturas (C)")
plt.legend(loc='upper right')
plt.show()
```
